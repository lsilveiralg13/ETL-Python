```sql
CREATE OR REPLACE VIEW gold.analistas.vw_producaocompletacomapontamento AS (

SELECT * 
FROM (
    -- Dados de 2025
    SELECT * 
    FROM gold.analistas.vw_producaocontagem2025

    UNION ALL

    -- Dados de 2026
    SELECT * 
    FROM gold.analistas.vw_producaocontagem2026
)
ORDER BY DataProducao DESC
);
```
